## 📌 Problem Statement

An E-commerce retail company operating **6 physical stores** across **5 regions** of Pakistan 
(Rawalpindi, Lahore, Karachi, Islamabad, Multan, Peshawar) has recorded **228 sales transactions** 
during **January–June 2024**. The raw transactional data is messy — it contains:

- Inconsistent text formatting (e.g., `" central"`, `"WEST"`, `"nida_iqbal"`, `"HASSAN SHEIKH"`)
- **Missing values** in `unit_price` (3), `order_date` (7), and `rating` (61 — ~27% of records)
- **Duplicate rows** and duplicated `order_id`s (8 exact duplicates)
- No revenue metric — only `quantity` and `unit_price` exist separately
- Store attributes (manager, city) stored in a **separate dimension table**


### 🎯 Project Objectives

| # | Objective |
|---|-----------|
| 1 | Clean and standardise the  dataset |
| 2 | Merge sales data with the store dimension table |
| 3 | Engineer business features: `total_amount`, `order_month`, `week_day`, `quarter`, `rating_bucket` |
| 4 | Perform **EDA** — univariate, bivariate, and pivot-based summaries |
| 5 | Visualise key insights (revenue by category, region, month) |
| 6 | Export a clean, analysis-ready dataset for Data Science Project |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

# Loading Data

```python
# Load the sale data
sales_df = pd.read_csv("sales_data.csv")
stores_df = pd.read_csv("stores.csv" )
```

```python
# Dimensions of each data
print("Sales shape :", sales_df.shape)
print("Stores shape:", stores_df.shape)
```

```python
# the top 5 rows for understand the data and their column
sales_df.head()
```

```python
# sales_df top  rows
stores_df.head()
```

# Data Audit and Cleaning

```python
# datatypes of all column
sales_df.dtypes
```

```python
# check Missing values
sales_df.isnull().sum()
```

```python
# unit_price: fill with median
median = sales_df['unit_price'].median()
sales_df['unit_price'] = sales_df['unit_price'].fillna(median )
```

```python
# unit_pirce: verify after fill 
sales_df['unit_price'].isna().sum()
```

```python
# order_date: → drop row (can't do time analysis) 
sales_df.dropna(subset=['order_date'], inplace=True)
```

```python
# rating: meadian by each category
median_by_cat = sales_df.groupby("category")["rating"].transform("median")
sales_df['rating'] = sales_df['rating'].fillna(median_by_cat)
```

```python
#rating: verify after fill by sum of null
sales_df['rating'].isna().sum()
```

```python
# lenth of df
before = len(sales_df)
sales_df.duplicated().sum() # check dupes
```

```python
# Drop duplicate values
sales_df.drop_duplicates(inplace=True)
```

```python
# order_id: for specific col
sales_df.drop_duplicates(subset=['order_id'] , keep='first' , inplace=True)
```

```python
print(f"Exact dupes removed: {before - len(sales_df)}")
```

```python
# check unique region
sales_df["region"].unique()
```

```python
sales_df.dtypes
```

```python
#  Fix dtypes
sales_df["order_date"] = pd.to_datetime(sales_df["order_date"], errors="coerce")
sales_df["unit_price"] = pd.to_numeric(sales_df["unit_price"], errors="coerce")
sales_df["rating"]     = pd.to_numeric(sales_df["rating"],     errors="coerce")
```

```python
# verify 
sales_df.dtypes
```

```python
# sales_df top5 rows
sales_df.head()
```

```python
# check unique region
sales_df["region"].unique()
```

```python
# region: strip whitespace, replace _ with space, standardize title case
sales_df["region"] = sales_df["region"].str.strip().str.title()
```

```python
# verify
sales_df["region"].unique()
```

```python
# Normalise text columns
# capital first letter, add _ and remove whitespace
sales_df["customer_name"] = (
    sales_df["customer_name"]
    .astype(str)
    .str.strip()
    .str.replace("_", " ", regex=False).str.title()
)
sales_df["product"]   = sales_df["product"].str.strip().str.title()
sales_df["category"]  = sales_df["category"].str.strip().str.title()
```

```python
# sales_df: we see strip whitespace, replace _ with space, standardize title case
sales_df.head()
```

## Merge

```python
#merge the sale $ store data 
final_df = sales_df.merge(stores_df , on= "store_id" , how = "left" , validate="many_to_one") # I will choose left join beacause data is transactional&dimensional 
```

```python
#top 5 rows of data it will help for understanding
final_df.head()
```

```python
#bottom 5 rows of data
final_df.tail()
```

```python
# summary of dataset
final_df.info(verbose=False)
```

# Feature Engineering

```python
# total amount with respect to quantity and price
final_df["total_amount"] = final_df["quantity"] * final_df["unit_price"]
final_df.head()
```

```python
#fianl_df: setting format of date and add order_year , week_day and quarter
final_df["order_month"]   = final_df["order_date"].dt.month
final_df["order_year"]    = final_df["order_date"].dt.year
final_df["week_day"]      = final_df["order_date"].dt.day_name()
final_df["quarter"]       = final_df["order_date"].dt.quarter
```

```python
final_df.head()
```

```python
# Balanced rating buckets: 1–2 low, 3 mid, 4–5 high
final_df["bucket"] = final_df["rating"].apply(
    lambda x: "unrated" if pd.isna(x)
    else "low" if x <= 2
    else "mid" if x <= 3
    else "high"
)
```

```python
# rating: Most frequent values
final_df["bucket"].value_counts()
```

# EDA

```python
# apply transpose
final_df.drop(columns=["order_id","order_date","order_year"]).describe().T
```

```python
# Revenue by category
rev_by_cat = final_df.groupby("category")["total_amount"].sum().sort_values()
rev_by_cat
```

```python
rev_by_cat.plot(kind="barh",
      color="steelblue",
      title="Revenue by Category")
```

```python
# Revenue by region
rev_by_reg = final_df.groupby("region")["total_amount"].sum().sort_values()
rev_by_reg
```

```python
rev_by_reg.plot(
    kind="bar",
    color="seagreen",
    title="Revenue by Region"
)
```

```python
# Monthly revenue trend
monthly_rev = final_df.groupby("order_month")["total_amount"].sum()
monthly_rev
```

```python
monthly_rev.plot(
    marker="o", color="crimson", title="Monthly Revenue Trend"
)
```

```python
# Rating distribution
order = ["low", "mid", "high"]  # adjust to your actual bucket names
counts = final_df["bucket"].value_counts().reindex(order)

counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Rating Bucket Share",
    ylabel="",
    startangle=90,
    counterclock=False,
)
```

# Aggregate & Summarize

```python
# Revenue with respect to category
rev_by_cat = final_df.groupby("category")["total_amount"].sum().sort_values(ascending=False)
```

```python
# Highest Revenue generated
rev_by_cat.idxmax()
```

```python
# Region × Category pivot
pivot_df = final_df.pivot_table(
    index="region", columns="category",
    values="total_amount", aggfunc="sum", fill_value=0
)
pivot_df
```

```python
# Region × Category pivot
final_df.groupby(["region", "category"]).agg(
    total_revenue=("total_amount", "sum"),
    avg_quantity=("quantity", "mean")
).round(2)
```

# Reshape 


```python
# Reshape 
melted_df = (
    pivot_df.reset_index()
    .melt(id_vars="region", var_name="category", value_name="total_amount")
)
melted_df.head(10)
```
